```python
import networkx as nx
import random
import matplotlib.pyplot as plt
import numpy

#add n nodes in the graph
def add_node(n):
    G=nx.Graph()
    G.add_nodes_from(range(n))
    return G

#adding random edge in the graph
def add_random_edge(G):
    v1=random.choice(list(G.nodes()))
    v2=random.choice(list(G.nodes()))
    if v1!=v2:
        G.add_edge(v1,v2)
    return G
#keeps adding random edges until the graph becomes connected
def add_till_connectivity(G):
    while(nx.is_connected(G)==False):
        G=add_random_edge(G)
    return G

#create a instance of entire process - it takes input of number of nodes and returns no of edges required for connectivity
def create_instance(n):
  G=add_node(n)
  G=add_till_connectivity(G)
  return G.number_of_edges()

#average it around 100 instance
def create_avg_instance(n):
  list1=[]
  for i in range(0,20):
    list1.append(create_instance(n))
  return numpy.average(list1)


#plotting the connectivituy
def plot_connectivity():
  x=[]
  y=[]
  i=10
  while(i<=100):
    x.append(i)
    y.append(create_avg_instance(i))
    i=i+10
    plt.xlabel("Number of Nodes")
    plt.ylabel("Number of Edges required for connecting the graph")
    plt.title("Emergence of Connectevity")
    plt.plot(x,y)
    x1=[]
    y1=[]
    i = 10
    while(i<=100):
      x1.append(i)
      y1.append(i*numpy.log(i))

    plt.plot(x1,y1)
    plt.show()
```
