```python
import networkx as nx
import matplotlib.pyplot as plt

H=nx.read_pajek("dotnetdataset.net")

print("Graph Information")
print("-----------------")

print("Number of Nodes :",H.number_of_nodes())
print("Number of Edges :",H.number_of_edges())

print("Is Directed :",nx.is_directed(H))

print("Density :",nx.density(H))

print("Average Clustering :",nx.average_clustering(H))

print("Connected Components :",nx.number_connected_components(H))

print("Nodes")
print(H.nodes())

print("Edges")
print(H.edges())

print("Degree of Nodes")
print(dict(H.degree()))

print("Degree Centrality")
print(nx.degree_centrality(H))

print("Betweenness Centrality")
print(nx.betweenness_centrality(H))

print("Closeness Centrality")
print(nx.closeness_centrality(H))

print("Shortest Path Example")
nodes=list(H.nodes())

print(nx.shortest_path(H,nodes[0],nodes[5]))

plt.figure(figsize=(10,8))

pos=nx.spring_layout(H)

nx.draw(H,
        pos,
        with_labels=True,
        node_size=700)

plt.show()
```
